```python
from dotenv import load_dotenv
from anthropic import Anthropic

load_dotenv()

client = Anthropic()
model ="claude-sonnet-5-5"
```

# Building a Flexible Chat Function

```python
#-------------------------------------------------------------------------------------------------------#
# Define a System Prompt
#-------------------------------------------------------------------------------------------------------#
system_prompt = """You are a patient math tutor. Do not directly answer a student's questions. Guide them to a solution step by step"""

#-------------------------------------------------------------------------------------------------------#
# Build a Flexible Chat Function
#-------------------------------------------------------------------------------------------------------#
# Function takes two arguments:
    # "messages": a required list of previous chat messages
    # "system": optional system prompt which defaults to non if non provided
def chat(messages, system = None):
    # Creates dictionary to store settings needed for the AI request
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }

    # "if system": checks if "system" variable has a value:
        # if system is None or empty string Python treats it as False and skips block
        # Claude's API doesn't accept "system=None"; need to conditionally include the system parameter only when it's provided.
    if system:
        # "params["system"]:" Dynamically adds a new key-value pair to dictionary; creates "system" key inside "params" and assigns "prompt" text to it
        params["system"] = system

    # Sending the Request
        # client.messages.create(...): Official Anthropic method to send a prompt to Claude.
        # "**params": Python’s keyword argument unpacking operator; unpacks "params" settings so it acts like you typed directly inside the function
    message = client.messages.create(**params)

    # Extracting and Joining the Text
        # "for block in message.content": Loops through every block returned in response payload
        # "if block.type == "text"": checks the block type; only looks at the blocks containing actual text response.
            # block.type: pulls out the raw text string inside those text blocks
            # "".join(...): merges all of individual text blocks into one continuous string and return
    return "".join(block.text for block in message.content if block.type == "text")

#-------------------------------------------------------------------------------------------------------#
# Define the conversation
#-------------------------------------------------------------------------------------------------------#

# messages = []: Creates an empty list named messages to hold the conversation flow.
messages = []

# messages.append(...): Adds a new message to the end of that list.
    # {"role": "user", "content": "..."}: Represents a single turn in the chat using a dictionary.
messages.append({"role": "user", "content": "How do I solve 5x + 2 = 3? for x?"})
```

```python
#-------------------------------------------------------------------------------------------------------#
# Call your helper with system prompt
#-------------------------------------------------------------------------------------------------------#
answer = chat(messages, system=system_prompt)
print(answer)

```

```python
#-------------------------------------------------------------------------------------------------------#
# Without helper without system prompt
#-------------------------------------------------------------------------------------------------------#

answer_2 = chat(messages)
print(answer_2)
```
