````python
from dotenv import load_dotenv
from anthropic import Anthropic

load_dotenv()

client = Anthropic()
model = "claude-sonnet-4-5"
````

````python
\{
    "source": ["aws.ec2"],
    "detail-type": ["EC2 Instance State-change Notification"],
    "detail": \{
        "state": ["running"]
    \}
\}
````

````python
#-------------------------------------------------------------------------------------------------------#
# Define a function for User Message Input
#-------------------------------------------------------------------------------------------------------#
def add_user_message(messages, text):
    messages.append({"role": "user", "content": text})

#-------------------------------------------------------------------------------------------------------#
# Define a function for User Message Input
#-------------------------------------------------------------------------------------------------------#
def add_assistant_message(messages, text):
    messages.append({"role": "assistant", "content": text})

#-------------------------------------------------------------------------------------------------------#
# Define a chat function
#-------------------------------------------------------------------------------------------------------#
def chat(messages, system = None, stop_sequences = None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }

    if system:
        params["system"] = system

     # "if stop_sequences": checks if "stop_sequences" variable has a value:
        # if there is None or an empty string Python treats it as False and skips block
        # Claude's API doesn't accept "stop_sequences = None"; need to conditionally include the system parameter only when it's provided.
    if stop_sequences:
        # If the check passes, this line creates (or updates) a "key" named "stop_sequences" inside the "params" dictionary and assigns the "value" of the "stop_sequences" variable to it.
        params["stop_sequences"] = stop_sequences

     # Sending the Request
    message = client.messages.create(**params)

    # Extracting and Joining the Text
    return "".join(block.text for block in message.content if block.type == "text")
````

````python
#-------------------------------------------------------------------------------------------------------#
# Define the conversation
#-------------------------------------------------------------------------------------------------------#

messages = []

add_user_message(messages, "Generate a very short event bridge rule as json")

add_assistant_message(messages, "```json")

text = chat(messages, stop_sequences=["```"])
````

````python
print(text)
````

````python
#-------------------------------------------------------------------------------------------------------#
# Convert text from Claude into Python data you can work with.
#-------------------------------------------------------------------------------------------------------#

# Loads Python’s built-in JSON module. It comes with Python, so nothing needs installing.
import json

# Clean up and parse the JSON
    # text.strip(): removes any spaces, blank lines, or newlines from the start and end of text
    # json.loads(...): Parses the string. The “s” stands for “string”: it reads JSON written as text and converts it into Python objects.
        # JSON objects ({...}) become dictionaries
        # JSON arrays ([...]) become lists.
    # clean_json: stores the result.
clean_json = json.loads(text.strip())

print(clean_json)
````

````python
# The difference matters because:
    # Before this line text is just a long string of characters, and you can’t ask a string for its "source" field.
    # After clean_json is a dictionary, so you can access its parts directly

# Print source: ['aws.ec2']
print(clean_json["source"])

# Print "detail-type": ['EC2 Instance State-change Notification']
print(clean_json["detail-type"])
````
