```python
from dotenv import load_dotenv
from anthropic import Anthropic

load_dotenv()

client = Anthropic()
model ="claude-sonnet-5-5"
```

# Building Helper Functions

```python
#-------------------------------------------------------------------------------------------------------#
# Define a function for User Message Input
    # Formats user input text into a structured dictionary
    # Appends user inputs into conversation history list.
#-------------------------------------------------------------------------------------------------------#

# Function accepts two arguments / inputs
    # "messages": list containing the conversation history
    # "text": a string representing the new user message
def add_user_message(messages, text):
    # Creates dictionary for user messages
    # "role": "user" tells the AI model that this specific message came from the human user
    user_message = {"role": "user", "content": text}
    # Appends dictionary with new user message
    messages.append(user_message)

#-------------------------------------------------------------------------------------------------------#
# Define a function for AI's response
    # Formats text into a structured dictionary
    # Appends responses into conversation history list.
#-------------------------------------------------------------------------------------------------------#

# Function accepts two arguments / inputs
    # "text": here represents new assistant response
def add_assistant_message(messages, text):
    #" role": "user" tells the AI model that this specific message came from the AI assistant
    assistant_message = {"role": "assistant", "content": text}
    # Appends dictionary with new assistant message
    messages.append(assistant_message)

#-------------------------------------------------------------------------------------------------------#
# Define a function that sends conversation history to an AI model
    # Use the Anthropic Claude API
    # Requests a response.
    # This serves as the bridge that connects the message history to the actual AI model, allowing it to read the conversation and generate the next reply.
#-------------------------------------------------------------------------------------------------------#

# Function only accepts one argument; "messages"
    # This is a complete list of dictionaries containing all user inputs and assistant responses.
def chat(messages):
    # client.messages.create:
        # Calls the official Anthropic Python SDK client to request a new message from Claude
        # Passes three key configurations to the API (model, max_tokens, messages)
    message = client.messages.create(
        # "model": Specifies which version of Claude to use
        # "max_tokens": Sets a hard limit on how long the AI's response can be
        # "messages": Feeds conversation history list directly into the API; provides model with context
        model = model,
        max_tokens=1000,
        messages = messages,
    )

    # print(message.content): temporarily shows every block in the response

    # Safely extracts and combines only the text-based parts of Anthropic Claude's API response into a single string
        # return: Sends the final, clean string back to whatever part of your program called the function.
        # "".join(...): Takes extracted text fragments from the loop and combines into one seamless string
        # "block.text": Extracts the actual string message written by Claude.
        # "for block in message.content": Loops through every block of data inside the API response object
        # "if block.type == "text"": Safety filter that checks each block and ensures the code only looks at blocks that actually contain spoken/written text
    return "".join(block.text for block in message.content if block.type == "text")
```

# Putting It All Together

```python
# Initializes an empty list
messages = []

# Calls function to add a new user request to conversation history list
add_user_message(messages,"Define quantum computing in one sentence")

# Executes API call to Anthropic Claude and saves the text response into a variable named
answer = chat(messages)

# Saves assistant response into conversation history list, closing the loop on this turn of the chat.
add_assistant_message(messages, answer)

# Adds follow-up request, "Write another sentence", to the conversation history list.
add_user_message(messages, "Write another sentence")

# Sends updated message history to Claude and saves the new follow-up response into a variable
final_answer = chat(messages)

print(answer)

print("-------------")

print(final_answer)
```
